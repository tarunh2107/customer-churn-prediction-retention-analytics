## 1. Import Libraries

```python
import pandas as pd
import numpy as np

# Set display options for full visibility
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
```

## 2. Identify Excel Sheet Name(s)

```python
file_path = '../data/raw/Telco_customer_churn.xlsx'
excel_file = pd.ExcelFile(file_path)
print("Available Sheet Name(s):", excel_file.sheet_names)
```

## 3. Load Raw Dataset

```python
sheet_name = 'Telco_Churn'
df = pd.read_excel(file_path, sheet_name=sheet_name)
print(f"Successfully loaded sheet: '{sheet_name}'")
```

## 4. First 5 Rows

```python
df.head(5)
```

## 5. Dataset Dimensions (Shape)

```python
n_rows, n_cols = df.shape
print(f"Number of Rows: {n_rows:,}")
print(f"Number of Columns: {n_cols}")
```

## 6. Column Names & Data Types

```python
print("--- Column Names and Data Types ---")
col_info = pd.DataFrame({
    'Column Name': df.columns,
    'Data Type': df.dtypes.values
})
col_info
```

```python
df.info()
```

## 7. Missing Values Analysis

```python
missing_counts = df.isnull().sum()
missing_percentage = (missing_counts / len(df) * 100).round(2)
missing_df = pd.DataFrame({
    'Missing Count': missing_counts,
    'Missing Percentage (%)': missing_percentage
})

print("Columns with Missing Values:")
missing_df[missing_df['Missing Count'] > 0]
```

## 8. Duplicate Rows Check

```python
num_duplicates = df.duplicated().sum()
print(f"Number of duplicate rows: {num_duplicates}")
```

## 9. Target / Churn Column Identification & Distribution

```python
# The target variables representing customer churn are:
# - 'Churn Label' ('No' / 'Yes')
# - 'Churn Value' (0 = Retained, 1 = Churned)

print("--- 'Churn Label' Value Counts ---")
print(df['Churn Label'].value_counts(dropna=False))

print("\n--- 'Churn Label' Proportions (%) ---")
print((df['Churn Label'].value_counts(normalize=True, dropna=False) * 100).round(2))

print("\n--- 'Churn Value' Value Counts ---")
print(df['Churn Value'].value_counts(dropna=False))
```

## 10. Summary of Initial Observations

- **Sheet Name:** `Telco_Churn`
- **Dataset Dimensions:** 7,043 rows, 33 columns.
- **Duplicates:** 0 duplicate rows.
- **Target Column:** `Churn Value` (0 = No Churn, 1 = Churn) / `Churn Label` (No / Yes).
- **Class Distribution:** 5,174 retained customers (73.46%) vs. 1,869 churned customers (26.54%).
- **Missing Values:** Only `Churn Reason` has 5,174 missing values (naturally corresponding to the 5,174 non-churned customers). No other column has explicit NaN values in raw form.
